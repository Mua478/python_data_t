```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
#print(df['job_posted_date'])
```

```python
df_filled = df.copy()#prevents df and df_filled from haveing the same id location(chnages will only occur to df_filled)
median_salary_hour = df_filled['salary_hour_avg'].median()
median_salary_year = df_filled['salary_year_avg'].median()

df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_hour)
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year)
# print(df_filled.loc[:,'salary_year_avg':'salary_hour_avg'])
df_filled.head(10)
```

```python
df_unique = df_filled.copy()
df_unique = df_unique.drop_duplicates()
df_unique = df_unique.drop_duplicates(subset = ['job_title','company_name'])
print(len(df_filled)-len(df_unique))

df.sample(10, random_state = 42) #same 10 random samples everytime
```

```python
df.pivot_table(values = 'salary_year_avg', index = 'job_country', columns = 'job_title_short', aggfunc = 'median').head(10)
```

```python
top_countries = df['job_country'].value_counts().head(6).index
df_job_country_salary = df.pivot_table(values = 'salary_year_avg', 
                index = 'job_country',
              columns = 'job_title_short',
              aggfunc = 'median' )

df_job_country_salary = df_job_country_salary.loc[top_countries]
job_titles = ['Data Analyst', 'Data Engineer', 'Data Scientist']
df_job_country_salary = df_job_country_salary[job_titles]
df_job_country_salary.plot(kind = 'line')
plt.title('Salary sorted by country')
plt.ylabel('Median salary($USD)')
plt.xlabel('Country')
plt.xticks(rotation = 45, ha = 'right')
plt.legend(title = 'Job Title')






```
